# Parameter Impact on the BH-BH Merger Rate

For each **parameter family** (e.g. common-envelope efficiency, stellar evolution tracks), this notebook shows — in one panel per family — how the local BH-BH merger rate changes across the model variations reported in the isolated-binary-evolution literature.

**Figure layout:**  
- One panel per parameter family.  
- One row per study that varies that parameter.  
- One scatter point per named sub-model of that study, connected by a thin horizontal line.  
- Fiducial / reference sub-model highlighted with a gold star (★).  
- Where a sub-model internally spans a range (multiple rate rows, e.g. metallicity bins), a shaded error bar shows [min, max].  

**Data source:** `Data_Mandel_and_Broekgaarden_2026/isolated-binary-evolution.csv`

```python
from pathlib import Path
import re
import warnings

import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from matplotlib import rc, rcParams

warnings.filterwarnings('ignore')

# ── Match the style of plot_rates.py ────────────────────────────────────────
rc("font", family="serif", weight="bold")
rc("text", usetex=True)
rcParams["text.latex.preamble"] = r"\usepackage{amsmath}" + "\n" + r"\boldmath"
rc("axes", linewidth=1.8)
rcParams.update({
    "xtick.major.size": 10, "ytick.major.size": 10,
    "xtick.minor.size": 6,  "ytick.minor.size": 6,
    "font.weight": "bold",
})

FONTSIZE       = 16
IBE_COLOR      = "#e07830"   # warm orange used for isolated-binary-evolution
FIDUCIAL_COLOR = "#f1c40f"   # gold for fiducial / reference sub-model
FIDUCIAL_EDGE  = "#b7950b"

_LATEX_ESCAPES = str.maketrans({"&": r"\&", "%": r"\%", "#": r"\#"})
def _tex(s: str) -> str:
    return s.translate(_LATEX_ESCAPES)
```

```python
# ── Load data ────────────────────────────────────────────────────────────────
DATA_DIR = Path("../Data_Mandel_and_Broekgaarden_2026")
DCO_TYPE = "BH-BH"    # change to "NS-NS" or "NS-BH" to inspect other DCO types

raw = pd.read_csv(DATA_DIR / "isolated-binary-evolution.csv", dtype=str).fillna("")
raw["rate"] = pd.to_numeric(raw["rate_Gpc3yr"], errors="coerce")
raw["year"] = pd.to_numeric(raw["year"],        errors="coerce")

# Keep only the requested DCO type and rows with a named sub-model
df = raw[(raw["compact_object_type"] == DCO_TYPE) & (raw["submodel"] != "")].copy()

print(f"Loaded {len(raw)} total rows; {len(df)} rows with named sub-models ({DCO_TYPE}).")
print(f"Studies with named sub-models: {df['study_key'].nunique()}")
```

```python
# ── Aggregate: one representative rate per (study_key, sub-model) ────────────
# Some studies store multiple rows per sub-model (e.g. different metallicity bins).
# We compute median / min / max to capture both the representative value and spread.
agg = (
    df.groupby(["study_key", "submodel"], as_index=False)
    .agg(
        label=("label",            "first"),
        year=("year",              "first"),
        parameter_family=("parameter_family", "first"),
        notes=("notes",            "first"),
        rate_repr=("rate",         "median"),   # representative rate
        rate_min=("rate",          "min"),
        rate_max=("rate",          "max"),
        n_rows=("rate",            "count"),
    )
)

# Keep only studies that have more than one named sub-model
n_sub = agg.groupby("study_key")["submodel"].transform("nunique")
agg = agg[n_sub > 1].copy()

# Fiducial / reference model detection (case-insensitive substring match)
_FIDUCIAL_PATTERNS = ["fiducial", "_ref", "ref_"]
def _is_fiducial(submodel: str) -> bool:
    s = submodel.lower()
    return any(p in s for p in ["fiducial"]) or s == "ref"

agg["is_fiducial"] = agg["submodel"].apply(_is_fiducial)

print(f"Multi-submodel studies: {agg['study_key'].nunique()}")
print(f"Total (study, sub-model) entries: {len(agg)}")
print(f"Fiducial models identified: {agg['is_fiducial'].sum()}")
agg[["study_key", "submodel", "rate_repr", "n_rows", "is_fiducial"]].head(20)
```

```python
# ── Simplified parameter family grouping ─────────────────────────────────────
# Each study is assigned to ONE simplified group based on its PRIMARY parameter
# (the text before the first '/'), with fallback to the full string.
# Edit CLASSIFY_RULES to adjust how parameter families map to panels.

CLASSIFY_RULES: list[tuple[str, str]] = [
    # (substring to match, simplified group label)
    ("common envelope",     r"Common Envelope ($\alpha_{\rm CE}$)"),
    ("ce efficiency",       r"Common Envelope ($\alpha_{\rm CE}$)"),
    ("mass transfer",       r"Mass Transfer ($\beta$, stability)"),
    ("natal kick",          r"Natal Kicks ($\sigma_{\rm kick}$)"),
    ("remnant mass",        r"Remnant Mass Prescription"),
    ("pisn",                r"Remnant Mass Prescription"),
    ("stellar track",       r"Stellar Evolution Tracks"),
    ("star formation",      r"Star Formation History"),
    ("metallicity",         r"Star Formation History"),
    ("initial condition",   r"Initial Conditions"),
    ("angular momentum",    r"Angular Momentum / Winds"),
]

def classify_param_family(pf: str) -> str:
    # Classify based on the PRIMARY parameter (text before the first '/'),
    # then fall back to the full string for secondary matches.
    primary = pf.lower().strip().split("/")[0].strip()
    for keyword, group in CLASSIFY_RULES:
        if keyword in primary:
            return group
    p = pf.lower().strip()
    for keyword, group in CLASSIFY_RULES:
        if keyword in p:
            return group
    return r"Multi-Parameter"

agg["group"] = agg["parameter_family"].apply(classify_param_family)

summary = (
    agg.drop_duplicates("study_key")[["study_key", "group", "year"]]
    .sort_values(["group", "year"])
    .reset_index(drop=True)
)
display(summary)
```

```python
# ── Build panel structure ─────────────────────────────────────────────────────
# Ordered list of panels; edit to reorder or exclude specific panels.
PANEL_ORDER = [
    r"Common Envelope ($\alpha_{\rm CE}$)",
    r"Mass Transfer ($\beta$, stability)",
    r"Remnant Mass Prescription",
    r"Natal Kicks ($\sigma_{\rm kick}$)",
    r"Stellar Evolution Tracks",
    r"Star Formation History",
    r"Initial Conditions",
    r"Angular Momentum / Winds",
    r"Multi-Parameter",
]

# Build panels dict: group → sorted list of study_keys
panels = {}
for g in PANEL_ORDER:
    g_df = agg[agg["group"] == g]
    if g_df.empty:
        continue
    studies = (
        g_df.drop_duplicates("study_key")
            .sort_values("year")["study_key"]
            .tolist()
    )
    panels[g] = {"studies": studies, "df": g_df}

print("Panels and studies:")
for g, p in panels.items():
    print(f"  [{g}]: {p['studies']}")
```

```python
# ── Drawing helpers ───────────────────────────────────────────────────────────

def draw_study_submodels(
    ax: plt.Axes,
    study_df: pd.DataFrame,
    y: float,
    color=IBE_COLOR,
    fiducial_color=FIDUCIAL_COLOR,
    fiducial_edge=FIDUCIAL_EDGE,
) -> None:
    """
    Draw one study row at vertical position `y`.
    Each named sub-model appears as one scatter point at its median rate.
    A horizontal line spans the full range of sub-model rates.
    If a sub-model has internal spread (multiple rate rows), add a subtle error bar.
    """
    sub = (
        study_df
        .drop_duplicates(subset=["submodel"])  # already aggregated, but safety check
        .sort_values("rate_repr")
        .dropna(subset=["rate_repr"])
    )
    if sub.empty:
        return

    rates = sub["rate_repr"].values

    # Horizontal connecting line spanning the full range of sub-model rates
    if len(rates) > 1:
        ax.plot(
            [rates.min(), rates.max()], [y, y],
            color=color, lw=3.0, zorder=3, alpha=0.55, solid_capstyle="butt",
        )

    # Scatter points (one per sub-model)
    for _, row in sub.iterrows():
        r = row["rate_repr"]
        if pd.isna(r):
            continue

        # Internal spread bar (e.g. metallicity bins within one sub-model)
        spread_lo = r - row["rate_min"]
        spread_hi = row["rate_max"] - r
        has_spread = (row["n_rows"] > 1) and (
            (spread_lo > 0.05 * r) or (spread_hi > 0.05 * r)
        )
        if has_spread:
            ax.errorbar(
                r, y,
                xerr=[[max(spread_lo, 0)], [max(spread_hi, 0)]],
                fmt="none", color=color, capsize=3,
                lw=1.5, alpha=0.45, zorder=4,
            )

        # Marker
        if row["is_fiducial"]:
            ax.scatter(
                r, y, s=280, marker="*",
                c=fiducial_color, edgecolors=fiducial_edge,
                linewidths=1.2, zorder=100,
            )
        else:
            ax.scatter(
                r, y, s=90, marker="o",
                c=[color], edgecolors="k",
                linewidths=0.7, zorder=50,
            )


def style_panel(ax, xmin, xmax, fontsize=FONTSIZE):
    """Apply common x-axis style to a panel axes."""
    ax.set_xscale("log")
    ax.set_xlim(xmin, xmax)
    ax.set_yticks([])
    ax.tick_params(which="major", length=8,  width=1.2, labelsize=fontsize - 2)
    ax.tick_params(which="minor", length=4,  width=1.0)
    for spine in ax.spines.values():
        spine.set_linewidth(1.2)
    # Vertical gridlines
    for xv in [1e-3, 1e-2, 1e-1, 1, 10, 1e2, 1e3, 1e4, 1e5]:
        if xmin <= xv <= xmax:
            ax.axvline(xv, lw=0.8, color="0.75", ls=":", zorder=0)
```

```python
# ── Main figure ───────────────────────────────────────────────────────────────
from matplotlib.transforms import blended_transform_factory

# Layout parameters
XMIN, XMAX     = 1e-2, 5e3   # rate axis limits (wider left range for label room)
FIG_WIDTH      = 18           # inches
ROW_H          = 0.44         # inches per study row
PANEL_HDR_H    = 0.55         # inches for panel header space
GAP_H          = 0.20         # inches between panels
TOP_MARGIN     = 0.70
BOT_MARGIN     = 0.80
LABEL_FONTSIZE = FONTSIZE - 3

# Compute panel heights
panel_heights = [
    len(panels[g]["studies"]) * ROW_H + PANEL_HDR_H
    for g in panels
]
total_h = sum(panel_heights) + (len(panels) - 1) * GAP_H + TOP_MARGIN + BOT_MARGIN

fig = plt.figure(figsize=(FIG_WIDTH, total_h))

# GridSpec: one row per panel (+ gap rows in between)
n_panels = len(panels)
gs_heights = []
for i, h in enumerate(panel_heights):
    gs_heights.append(h)
    if i < n_panels - 1:
        gs_heights.append(GAP_H)

gs = gridspec.GridSpec(
    nrows=len(gs_heights), ncols=1,
    height_ratios=gs_heights,
    left=0.30,    # leave 30% left margin for study labels
    right=0.97,
    top=1 - TOP_MARGIN / total_h,
    bottom=BOT_MARGIN / total_h,
    hspace=0,
)

axes = []
for panel_idx, (group, panel) in enumerate(panels.items()):
    gs_row  = panel_idx * 2   # every other row is a gap
    ax      = fig.add_subplot(gs[gs_row, 0])
    style_panel(ax, XMIN, XMAX)
    axes.append(ax)

    studies   = panel["studies"]
    n_studies = len(studies)
    ax.set_ylim(-n_studies - 0.5, 0.5)

    # Blended transform: x in axes fraction, y in data coords
    trans_label = blended_transform_factory(ax.transAxes, ax.transData)

    for study_idx, sk in enumerate(studies):
        y        = -study_idx
        study_df = panel["df"][panel["df"]["study_key"] == sk]

        draw_study_submodels(ax, study_df, y)

        # Study label in the left margin (outside the data area)
        label_text = _tex(study_df["label"].iloc[0])
        ax.text(
            -0.02, y, label_text,
            ha="right", va="center",
            fontsize=LABEL_FONTSIZE, color="0.15",
            transform=trans_label,
        )

        # Light separator between rows
        if study_idx < n_studies - 1:
            ax.axhline(-study_idx - 0.5, color="0.88", lw=0.7, zorder=0)

    # Panel title: right side of panel in data coords
    ax.text(
        0.99, 0.50,
        group,
        ha="right", va="center",
        fontsize=FONTSIZE - 1, color=IBE_COLOR, fontweight="bold",
        transform=ax.transAxes,
    )

    # Hide x-tick labels on all but the bottom panel
    if panel_idx < n_panels - 1:
        ax.tick_params(labelbottom=False)

# Shared x-label on bottom panel
axes[-1].set_xlabel(
    r"$\mathbf{BH\text{-}BH\ local\ merger\ rate\ density}\ [\mathrm{Gpc}^{-3}\,\mathrm{yr}^{-1}]$",
    fontsize=FONTSIZE + 2, labelpad=6,
)

# Legend
from matplotlib.lines import Line2D
legend_elements = [
    Line2D([0], [0], marker="*", color="w", markerfacecolor=FIDUCIAL_COLOR,
           markeredgecolor=FIDUCIAL_EDGE, markersize=14, label="Fiducial / reference model"),
    Line2D([0], [0], marker="o", color="w", markerfacecolor=IBE_COLOR,
           markeredgecolor="k", markersize=9, label="Parameter variant"),
    Line2D([0], [0], color=IBE_COLOR, lw=3, alpha=0.55, label="Rate range"),
]
axes[0].legend(
    handles=legend_elements,
    loc="upper left", fontsize=FONTSIZE - 3,
    framealpha=0.9, edgecolor="0.8",
)

plt.savefig("parameter_impact_BHBH.pdf", bbox_inches="tight", dpi=200)
plt.savefig("parameter_impact_BHBH.png", bbox_inches="tight", dpi=200)
print("Saved: parameter_impact_BHBH.pdf / .png")
plt.show()
```

```python
# ── Optional: per-panel close-up figures ─────────────────────────────────────
# Uncomment to generate one figure per parameter family.

def make_single_panel(group: str, save: bool = False):
    """Create a standalone figure for one parameter family panel."""
    if group not in panels:
        print(f"Group not found: {group}")
        return

    panel  = panels[group]
    studies = panel["studies"]
    n_studies = len(studies)

    fig_h = n_studies * ROW_H + 1.5
    fig, ax = plt.subplots(figsize=(FIG_WIDTH * 0.7, fig_h))

    style_panel(ax, XMIN, XMAX)
    ax.set_ylim(-n_studies - 0.5, 0.5)

    for study_idx, sk in enumerate(studies):
        y        = -study_idx
        study_df = panel["df"][panel["df"]["study_key"] == sk]
        draw_study_submodels(ax, study_df, y)

        label_text = _tex(study_df["label"].iloc[0])
        ax.text(XMIN * 1.05, y, label_text,
                ha="left", va="center", fontsize=LABEL_FONTSIZE, color="0.15")
        ax.axhline(-study_idx - 0.5, color="0.88", lw=0.7, zorder=0)

    ax.set_title(group, fontsize=PANEL_FONTSIZE + 2, color=IBE_COLOR, pad=8)
    ax.set_xlabel(
        r"$\mathbf{BH\text{-}BH\ merger\ rate}\ [\mathrm{Gpc}^{-3}\,\mathrm{yr}^{-1}]$",
        fontsize=FONTSIZE,
    )

    plt.tight_layout()
    if save:
        slug = re.sub(r"[^\w]", "_", group).strip("_").lower()
        plt.savefig(f"parameter_impact_BHBH_{slug}.pdf", bbox_inches="tight", dpi=200)
        plt.savefig(f"parameter_impact_BHBH_{slug}.png", bbox_inches="tight", dpi=200)
    plt.show()


# ── Uncomment any of these to get standalone panel figures: ─────────────────
# make_single_panel(r"Common Envelope ($\alpha_{\rm CE}$)", save=True)
# make_single_panel(r"Mass Transfer ($\beta$, stability)", save=True)
# make_single_panel(r"Stellar Evolution Tracks", save=True)
```
